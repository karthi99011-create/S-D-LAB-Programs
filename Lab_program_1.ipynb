```python
#numpy: basic statistical computation
import numpy as np
marks=np.array([65,72,81,55,90])
print("Mean:",marks.mean())
print("Maximum:",marks.max())
print("Minimum:",marks.min())
print("std:",marks.std())
```

```python
#pandas:Working with tabular data
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
data={"Name":["A","B","C","D","E"],"MARKS":[65,72,81,55,90]}

df=pd.DataFrame(data)
print(df)
print(df["MARKS"].mean())

#Visualization: Turn numbers into patterns

plt.bar(df["Name"],df["MARKS"])
plt.xlabel("student")
plt.ylabel("marks")
plt.title("student marks")
plt.show()

plt.hist(df["MARKS"])
plt.xlabel("marks")
plt.ylabel("Frequency")
plt.title("Distribution of marks")
plt.show()
```

```python
# Experiment1:Probability throught Simulation(Coin Toss:one simulation)
import numpy as np
result=np.random.choice(["Head","Tail"],size=10)
print(result)

#Estimate probability
result=np.random.choice(["Head","Tail"],size=10)
heads=np.sum(result=="Head")
print("Heads:",heads)
print("probability of Head:",heads/10)
```

```python
#   Visualize the coin  toss
counts=[np.sum(result=="Head"),
        np.sum(result=="Tail")]

plt.bar(["Head","Tail"],counts)
plt.xlabel("outcome")
plt.ylabel("count")
plt.title("coin toss simulation")
plt.show()
```
